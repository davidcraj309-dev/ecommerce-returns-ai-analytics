```python
import os

path = "C:\\Users\\Yesu\\OneDrive\\Documents\\Desktop"
file_name = "ecommerce_orders.log.txt"




clean_category = []
clean_status = []
clean_issue = []
clean_Return = []
clean_amount = []

with open(os.path.join(path,file_name)) as data_cleaing:
    for each_line in data_cleaing:
        cleaned_data =  each_line.strip().split(" ")


        category = cleaned_data[3].split("=")[1]
        status = cleaned_data[4].split("=")[1]
        issue = cleaned_data[5].split("=")[1]
        Return = cleaned_data[6].split("=")[1]
        amount = cleaned_data[7].split("=")[1]



    
        clean_category.append(category)
        clean_status.append(status)
        clean_issue.append(issue)
        clean_Return.append(Return)

        




print(clean_category)
print(clean_status)
print(clean_issue)
print(clean_Return)


```

```python
def frequency(cleaned_data):
    mydict = {}

    for each in cleaned_data:
        if each not in mydict:
            mydict[each] = 1
        else:
            mydict[each] += 1

    return mydict



rest1 = frequency(clean_category)
rest2 = frequency(clean_status)
rest3 = frequency(clean_issue)
rest4 = frequency(clean_Return)


print("Category Frequency:",rest1)
print("Status Frequency:",rest2)
print("Issue Frequency:",rest3)
print("Return Frequency:",rest4)
```

```python
import csv
with open("category_frequency.csv","w",newline="") as my_Category:
    writer = csv.writer(my_Category)
    writer.writerow(["Category","Count"])

    for Category ,Count in rest1.items():
        writer.writerow([Category,Count])
```

```python
import csv
with open("Status_frequency.csv","w",newline="") as my_Status:
    writer = csv.writer(my_Status)
    writer.writerow(["Status","Count"])

    for Status ,Count in rest2.items():
        writer.writerow([Status ,Count])
```

```python
import csv
with open("Issue_frequency.csv","w",newline="") as my_Issue:
    writer = csv.writer(my_Issue)
    writer.writerow(["Issue","Count"])

    for Issue  ,Count in rest3.items():
        writer.writerow([Issue  ,Count])
```

```python
import csv
with open("Return_frequency.csv","w",newline="") as my_Return:
    writer = csv.writer(my_Return)
    writer.writerow(["Return","Count"])

    for Return ,Count in rest4.items():
        writer.writerow([Return ,Count])
```
